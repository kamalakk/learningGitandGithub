```python
plot_data = df[
	['year_introduced', 'speed_mph', 'height_value',
	 'Inversions_clean', 'Gforce_clean', 'Type_Main']
].copy()

numeric_columns = [
	'year_introduced', 'speed_mph', 'height_value',
	'Inversions_clean', 'Gforce_clean'
]

plot_data[numeric_columns] = plot_data[numeric_columns].apply(
	pd.to_numeric, errors='coerce'
)

sns.pairplot(
	data=plot_data,
	vars=numeric_columns,
	hue='Type_Main'
)
plt.show()
```

```python
df_corr=df[['year_introduced', 'speed_mph', 'height_value', 'Inversions_clean', 'Gforce_clean']].dropna().corr()
```

```python
sns.heatmap(df_corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.show()
```
